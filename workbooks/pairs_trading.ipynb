```python
correlation_min = 0.7
correlation_max = 0.9
upper_triangle = np.triu(
    np.ones(returns_correlation.shape, dtype=bool),
    k=1,
)

correlation_pairs = (
    returns_correlation.where(upper_triangle)
    .stack()
    .loc[lambda correlations: (correlations >= correlation_min) & (correlations <= correlation_max)]
    .sort_values(ascending=False)
)

correlation_pairs
```

```python
import os
from joblib import Parallel, delayed

base_corr = 0.65 # I've test .5-.65
high_corr = 0.8
min_dur = 5
rolling_window = 20
n_jobs = min(8, os.cpu_count() or 2)

symbol_names = returns_correlation.columns.to_numpy()
correlation_values = returns_correlation.to_numpy(copy=False)
returns_values = returns_df.to_numpy(dtype='float64', copy=False)


def iter_candidate_pairs():
    """Yield unique column-index pairs meeting the base correlation."""
    for row_index in range(len(symbol_names) - 1):
        qualifying_offsets = np.flatnonzero(
            correlation_values[row_index, row_index + 1:] >= base_corr
        )
        for offset in qualifying_offsets:
            yield row_index, row_index + 1 + offset


def evaluate_pair(
    row_index,
    column_index,
    values,
    window,
    correlation_threshold,
    minimum_duration,
):
    """Evaluate one pair in a worker and return retained correlation values."""
    pair_values = values[:, [row_index, column_index]]
    pair_values = pair_values[~np.isnan(pair_values).any(axis=1)]

    if len(pair_values) < window:
        return None

    rolling_correlation = pd.Series(pair_values[:, 0]).rolling(
        window=window,
        min_periods=window,
    ).corr(pd.Series(pair_values[:, 1]))

    above_threshold = rolling_correlation.gt(correlation_threshold).to_numpy()
    if len(above_threshold) < minimum_duration:
        return None

    consecutive_counts = np.convolve(
        above_threshold,
        np.ones(minimum_duration, dtype=np.int16),
        mode='valid',
    )
    if not np.any(consecutive_counts >= minimum_duration):
        return None

    return row_index, column_index, rolling_correlation.to_numpy()


# Count without materializing the potentially large candidate iterable.
candidate_count = sum(
    np.count_nonzero(correlation_values[row_index, row_index + 1:] >= base_corr)
    for row_index in range(len(symbol_names) - 1)
)
pair_iterable = iter_candidate_pairs()

# Joblib memory-maps the shared returns array for worker processes. Only pair
# indices and small retained results are transferred for each task.
tasks = (
    delayed(evaluate_pair)(
        row_index,
        column_index,
        returns_values,
        rolling_window,
        high_corr,
        min_dur,
    )
    for row_index, column_index in pair_iterable
)
results = Parallel(
    n_jobs=n_jobs,
    prefer='processes',
    return_as='generator',
    pre_dispatch='2*n_jobs',
    max_nbytes='1M',
    mmap_mode='r',
)(tasks)

kept_pairs = {}
progress = tqdm(
    results,
    total=candidate_count,
    desc='Evaluating pairs',
    unit='pair',
)

for result in progress:
    if result is not None:
        row_index, column_index, rolling_correlation = result
        symbol_a = symbol_names[row_index]
        symbol_b = symbol_names[column_index]
        pair_data = returns_df.iloc[:, [row_index, column_index]].dropna().copy()
        pair_data['rolling_correlation'] = rolling_correlation
        kept_pairs[(symbol_a, symbol_b)] = pair_data

    progress.set_postfix(kept=len(kept_pairs), refresh=False)

progress.close()
print(f'Kept {len(kept_pairs):,} of {candidate_count:,} candidate pairs.')
```

```python
import plotly.graph_objects as go


def plot_intraday_ohlcv(symbols, date_range, timeframe):
    """Fetch intraday data and overlay each symbol's cumulative close return.

    Parameters
    ----------
    symbols : list[str]
        Symbols to chart.
    date_range : tuple[str | datetime, str | datetime]
        Inclusive start and end dates.
    timeframe : int
        Number of minutes per bar.
    """
    symbols = list(dict.fromkeys(symbol.upper() for symbol in symbols))
    if not symbols:
        raise ValueError('symbols must contain at least one symbol.')
    if len(date_range) != 2:
        raise ValueError('date_range must contain a start date and an end date.')
    if not isinstance(timeframe, int) or timeframe <= 0:
        raise ValueError('timeframe must be a positive integer number of minutes.')

    start_date, end_date = map(pd.Timestamp, date_range)
    if start_date > end_date:
        raise ValueError('The start date must not be after the end date.')

    intraday_data = intraday_import(
        wl=symbols,
        from_date=start_date.strftime('%Y-%m-%d'),
        to_date=end_date.strftime('%Y-%m-%d'),
        timespan='minute',
        multiplier=timeframe,
        market_open_only=True,
    )

    figure = go.Figure()
    for symbol in symbols:
        price_data = intraday_data.get(symbol)
        if price_data is None or price_data.empty:
            print(f'No intraday data returned for {symbol}.')
            continue

        close = price_data['Close'].dropna()
        if close.empty:
            print(f'No close-price data returned for {symbol}.')
            continue

        cumulative_return = close.div(close.iloc[0]).sub(1)
        figure.add_trace(
            go.Scatter(
                x=cumulative_return.index,
                y=cumulative_return,
                mode='lines',
                name=symbol,
            )
        )

    figure.update_layout(
        title=f'Intraday Cumulative Returns — {timeframe}-Minute Bars',
        template='plotly_white',
        height=600,
        hovermode='x unified',
        xaxis_title='Timestamp',
        yaxis_title='Cumulative Return',
        legend_title='Symbol',
    )
    figure.update_xaxes(
        rangebreaks=[
            dict(bounds=[16, 9.5], pattern='hour'),
            dict(bounds=['sat', 'mon']),
        ]
    )
    figure.update_yaxes(tickformat='.1%')
    figure.show()

    return figure


# Example:
intraday_figure = plot_intraday_ohlcv(
    symbols=['NBIS', 'CRWV'],
    date_range=('2026-08-12', '2026-08-18'),
    timeframe=1,
)
intraday_figure
```
